# Cell 1: Imports & Setup

```python
import pyspark.sql.functions as F
from pyspark.sql.types import *

# Setup para Community Edition
catalog = "workspace"
schema = "ecommerce_dq"

print(f"📊 Usando: {catalog}.{schema}")

# Criar schema
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}`")
print(f"✅ Schema criado!")

# Verificar
spark.sql(f"SHOW SCHEMAS IN {catalog}").show()
```

# Cell 2: Create Bronze Tables from Sample Data


```python
# Cell 2: Create Bronze Tables from Sample Data
def create_bronze_products():
    """Create products bronze table"""
    data = [
        (1, "Laptop", "ELECTRONICS", 999.99, 4.5),
        (2, "Mouse", "ELECTRONICS", 29.99, 4.2),
        (3, "Keyboard", "ELECTRONICS", 79.99, 4.3),
        (4, "Monitor", "ELECTRONICS", 299.99, 4.4),
        (5, "Desk Chair", "FURNITURE", 199.99, 4.1),
    ]
    
    schema = StructType([
        StructField("id", IntegerType()),
        StructField("title", StringType()),
        StructField("category", StringType()),
        StructField("price", DoubleType()),
        StructField("rating", DoubleType()),
    ])
    
    df = spark.createDataFrame(data, schema)
    
    # Add metadata
    df = df.withColumn("ingestion_date", F.current_date())
    df = df.withColumn("ingestion_timestamp", F.current_timestamp())
    df = df.withColumn("source", F.lit("s3://ecommerce-data-platform-datalake-211854352436/bronze/products"))
    
    # Save as Delta
    table_name = f"{catalog}.{schema}.bronze_products"
    df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(table_name)
    
    print(f"✅ Created: {table_name} ({df.count()} rows)")
    return df
```

# Cell 4: Create Bronze Orders


```python

def create_bronze_orders():
    """Create orders bronze table"""
    data = [
        (1, 1, "2026-05-15", "completed", 999.99),
        (2, 2, "2026-05-16", "processing", 29.99),
        (3, 1, "2026-05-17", "completed", 199.99),
        (4, 3, "2026-05-18", "shipped", 79.99),
        (5, 4, "2026-05-19", "completed", 299.99),
    ]
    
    schema = StructType([
        StructField("id", IntegerType()),
        StructField("customer_id", IntegerType()),
        StructField("order_date", StringType()),
        StructField("status", StringType()),
        StructField("total_amount", DoubleType()),
    ])
    
    df = spark.createDataFrame(data, schema)
    df = df.withColumn("order_date", F.to_date(F.col("order_date")))
    
    # Add metadata
    df = df.withColumn("ingestion_date", F.current_date())
    df = df.withColumn("ingestion_timestamp", F.current_timestamp())
    df = df.withColumn("source", F.lit("s3://ecommerce-data-platform-datalake-211854352436/bronze/orders"))
    
    # Save as Delta
    table_name = f"{catalog}.{schema}.bronze_orders"
    df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(table_name)
    
    print(f"✅ Created: {table_name} ({df.count()} rows)")
    return df
```

# Cell 5: Create Bronze Events


```python
# CELL 2: Create Bronze Tables (CORRIGIDO)

def create_bronze_products():
    """Create products bronze table"""
    data = [
        (1, "Laptop", "ELECTRONICS", 999.99, 4.5),
        (2, "Mouse", "ELECTRONICS", 29.99, 4.2),
        (3, "Keyboard", "ELECTRONICS", 79.99, 4.3),
        (4, "Monitor", "ELECTRONICS", 299.99, 4.4),
        (5, "Desk Chair", "FURNITURE", 199.99, 4.1),
    ]
    
    product_schema = StructType([
        StructField("id", IntegerType()),
        StructField("title", StringType()),
        StructField("category", StringType()),
        StructField("price", DoubleType()),
        StructField("rating", DoubleType()),
    ])
    
    df = spark.createDataFrame(data, product_schema)
    df = df.withColumn("ingestion_date", F.current_date())
    df = df.withColumn("ingestion_timestamp", F.current_timestamp())
    df = df.withColumn("source", F.lit("s3://ecommerce-data-platform-datalake-211854352436/bronze/products"))
    
    table_name = f"{catalog}.{schema}.bronze_products"
    df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(table_name)
    
    print(f"✅ Created: {table_name} ({df.count()} rows)")
    return df

def create_bronze_customers():
    """Create customers bronze table"""
    data = [
        (1, "john@email.com", "John", "Smith", "New York", "10001"),
        (2, "jane@email.com", "Jane", "Doe", "Los Angeles", "90001"),
        (3, "bob@email.com", "Bob", "Johnson", "Chicago", "60601"),
        (4, "alice@email.com", "Alice", "Williams", "Houston", "77001"),
        (5, "charlie@email.com", "Charlie", "Brown", "Phoenix", "85001"),
    ]
    
    customer_schema = StructType([
        StructField("id", IntegerType()),
        StructField("email", StringType()),
        StructField("firstname", StringType()),
        StructField("lastname", StringType()),
        StructField("city", StringType()),
        StructField("zipcode", StringType()),
    ])
    
    df = spark.createDataFrame(data, customer_schema)
    df = df.withColumn("ingestion_date", F.current_date())
    df = df.withColumn("ingestion_timestamp", F.current_timestamp())
    df = df.withColumn("source", F.lit("s3://ecommerce-data-platform-datalake-211854352436/bronze/customers"))
    
    table_name = f"{catalog}.{schema}.bronze_customers"
    df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(table_name)
    
    print(f"✅ Created: {table_name} ({df.count()} rows)")
    return df

def create_bronze_orders():
    """Create orders bronze table"""
    data = [
        (1, 1, "2026-05-15", "completed", 999.99),
        (2, 2, "2026-05-16", "processing", 29.99),
        (3, 1, "2026-05-17", "completed", 199.99),
        (4, 3, "2026-05-18", "shipped", 79.99),
        (5, 4, "2026-05-19", "completed", 299.99),
    ]
    
    order_schema = StructType([
        StructField("id", IntegerType()),
        StructField("customer_id", IntegerType()),
        StructField("order_date", StringType()),
        StructField("status", StringType()),
        StructField("total_amount", DoubleType()),
    ])
    
    df = spark.createDataFrame(data, order_schema)
    df = df.withColumn("order_date", F.to_date(F.col("order_date")))
    df = df.withColumn("ingestion_date", F.current_date())
    df = df.withColumn("ingestion_timestamp", F.current_timestamp())
    df = df.withColumn("source", F.lit("s3://ecommerce-data-platform-datalake-211854352436/bronze/orders"))
    
    table_name = f"{catalog}.{schema}.bronze_orders"
    df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(table_name)
    
    print(f"✅ Created: {table_name} ({df.count()} rows)")
    return df

def create_bronze_events():
    """Create events bronze table"""
    data = [
        ("evt_001", "page_view", "2026-06-02 10:00:00", 1, "desktop"),
        ("evt_002", "product_view", "2026-06-02 10:05:00", 2, "mobile"),
        ("evt_003", "add_to_cart", "2026-06-02 10:10:00", 3, "desktop"),
        ("evt_004", "purchase", "2026-06-02 10:15:00", 1, "mobile"),
        ("evt_005", "page_view", "2026-06-02 10:20:00", 4, "tablet"),
    ]
    
    event_schema = StructType([
        StructField("event_id", StringType()),
        StructField("event_type", StringType()),
        StructField("timestamp", StringType()),
        StructField("user_id", IntegerType()),
        StructField("device", StringType()),
    ])
    
    df = spark.createDataFrame(data, event_schema)
    df = df.withColumn("timestamp", F.to_timestamp(F.col("timestamp")))
    df = df.withColumn("ingestion_date", F.current_date())
    df = df.withColumn("ingestion_timestamp", F.current_timestamp())
    df = df.withColumn("source", F.lit("s3://ecommerce-data-platform-datalake-211854352436/bronze/events"))
    
    table_name = f"{catalog}.{schema}.bronze_events"
    df.write.format("delta").mode("overwrite").option("mergeSchema", "true").saveAsTable(table_name)
    
    print(f"✅ Created: {table_name} ({df.count()} rows)")
    return df
```
